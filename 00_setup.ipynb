```python
%sql
-- Catálogo = a "pasta-mãe" do projeto
CREATE CATALOG IF NOT EXISTS cvm;

-- Schemas = as camadas da arquitetura medalhão
CREATE SCHEMA IF NOT EXISTS cvm.bronze COMMENT 'Dados crus da CVM, como vieram';
CREATE SCHEMA IF NOT EXISTS cvm.silver COMMENT 'Dados limpos e tipados';
CREATE SCHEMA IF NOT EXISTS cvm.gold   COMMENT 'Indicadores prontos para análise';

-- Volume = onde ficam os arquivos (zip/csv) baixados da CVM
CREATE VOLUME IF NOT EXISTS cvm.bronze.arquivos_cvm;

SHOW SCHEMAS IN cvm;
```

```python
import requests

url = "https://dados.cvm.gov.br/dados/CIA_ABERTA/DOC/DFP/DADOS/"
try:
    r = requests.get(url, timeout=20)
    print("Status:", r.status_code)
    print("Acesso OK!" if r.ok else "Respondeu, mas com erro")
except Exception as e:
    print("Bloqueado:", type(e).__name__, "-", str(e)[:200])
```
